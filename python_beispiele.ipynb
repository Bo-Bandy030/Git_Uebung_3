## 2. CSV-Dateien lesen und schreiben

Wir erstellen einen kleinen DataFrame, speichern ihn als CSV und lesen die Datei wieder ein. Der temporäre Ordner wird beim Schließen der Python-Sitzung automatisch entfernt.

```python
from pathlib import Path
from tempfile import TemporaryDirectory

arbeitsordner = TemporaryDirectory()
csv_pfad = Path(arbeitsordner.name) / "umsatz.csv"

umsatz = pd.DataFrame({
    "Monat": ["Januar", "Februar", "März", "April", "Mai", "Juni"],
    "Region": ["Nord", "Süd", "Nord", "Süd", "Nord", "Süd"],
    "Umsatz": [1200, 950, 1350, 1100, 1500, 1250],
})
umsatz.to_csv(csv_pfad, index=False)

umsatz = pd.read_csv(csv_pfad)
umsatz
```

## 3. Datenanalyse mit pandas

`head()` zeigt die ersten Zeilen. Mit einer Bedingung filtern wir Datensätze; `groupby()` fasst Werte nach einer Kategorie zusammen.

```python
print("Erste Zeilen:")
display(umsatz.head())

print("Umsätze ab 1.200:")
display(umsatz[umsatz["Umsatz"] >= 1200])

print("Durchschnittlicher Umsatz je Region:")
display(umsatz.groupby("Region")["Umsatz"].mean().rename("Durchschnitt").reset_index())
```

## 4. Visualisierung mit Matplotlib

Die folgenden Diagramme verwenden dieselben Umsatzdaten: ein Liniendiagramm für den Verlauf und ein Balkendiagramm für die Summe je Region.

```python
fig, achsen = plt.subplots(1, 2, figsize=(12, 4))

achsen[0].plot(umsatz["Monat"], umsatz["Umsatz"], marker="o", color="teal")
achsen[0].set_title("Umsatz im Zeitverlauf")
achsen[0].set_ylabel("Umsatz")
achsen[0].tick_params(axis="x", rotation=35)

umsatz_je_region = umsatz.groupby("Region")["Umsatz"].sum()
achsen[1].bar(umsatz_je_region.index, umsatz_je_region.values, color=["#e07a5f", "#3d8b8b"])
achsen[1].set_title("Umsatz je Region")
achsen[1].set_ylabel("Umsatz gesamt")

fig.tight_layout()
plt.show()
```

## 5. Berechnungen mit NumPy

NumPy-Arrays erlauben elementweise Operationen ohne Schleifen. Hier berechnen wir Rabatte und statistische Kennzahlen für mehrere Preise.

```python
preise = np.array([19.99, 34.50, 12.00, 48.00])
preise_mit_rabatt = preise * 0.9

print("Preise:", preise)
print("Preise mit 10 % Rabatt:", np.round(preise_mit_rabatt, 2))
print("Mittelwert:", round(np.mean(preise), 2))
print("Minimum und Maximum:", np.min(preise), np.max(preise))
```

## 6. Eigene Funktionen erstellen und ausführen

Funktionen bündeln wiederverwendbare Logik. Die Funktion unten berechnet den Preis nach einem prozentualen Rabatt und prüft, dass der Rabatt zwischen 0 und 100 Prozent liegt.

```python
def preis_nach_rabatt(preis, rabatt_prozent):
    if not 0 <= rabatt_prozent <= 100:
        raise ValueError("Der Rabatt muss zwischen 0 und 100 liegen.")
    return preis * (1 - rabatt_prozent / 100)

preis_nach_rabatt(80, 15)
```

## 7. Markdown und mathematische Formeln

Markdown-Zellen eignen sich für Erklärungen und mathematische Notation. Der arithmetische Mittelwert einer Folge mit $n$ Werten ist

$$
\bar{x} = \frac{1}{n} \sum_{i=1}^{n} x_i.
$$

Für die Preise aus dem NumPy-Beispiel ergibt sich damit folgender Mittelwert:

```python
mittelwert = np.sum(preise) / len(preise)
print(f"Mittelwert nach der Formel: {mittelwert:.2f}")
```
